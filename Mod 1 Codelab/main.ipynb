```python
# Mahligai Arsya Nanda | NIM 202410370110056 | Fungsional I
def total_sum(n):
    if n == 0:                 # base case
        print("  total_sum(0) -> base case, return 0")
        return 0
    print(f"  total_sum({n}) -> {n} + total_sum({n-1})")
    result = n + total_sum(n - 1)
    print(f"  total_sum({n}) returns {result}")
    return result

print("Result :", total_sum(3))
```
