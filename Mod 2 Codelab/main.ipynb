```python
# Mahligai Arsya Nanda | NIM 202410370110056 | Fungsional I
# the data that we have
exam_scores = [70, 80, 85]
print("Original exam_scores:", exam_scores)
```

```python
# Mahligai Arsya Nanda | NIM 202410370110056 | Fungsional I
def add_bonus2(bonus):
    for i in range(len(exam_scores)):
        exam_scores[i] = exam_scores[i] + bonus
    return exam_scores
```

```python
# Mahligai Arsya Nanda | NIM 202410370110056 | Fungsional I
print("Before:", exam_scores)

result = add_bonus2(5)

print("New result:", result)
print("Original exam_scores be:", exam_scores)
```

```python
# Mahligai Arsya Nanda | NIM 202410370110056 | Fungsional I
server_logs = [
    "INFO: User login successful",
    "ERROR: Database connection failed",
    "INFO: Page rendered in 25ms",
    "ERROR: Unauthorized access attempt",
    "INFO: User logout successful"
]
```

```python
# Mahligai Arsya Nanda | NIM 202410370110056 | Fungsional I
def filter_emergency_log(log_list):
    for log in log_list:
        if log.startswith("ERROR"):
            yield log
```

```python
# Mahligai Arsya Nanda | NIM 202410370110056 | Fungsional I
error_bucket = filter_emergency_log(server_logs)

print(next(error_bucket))
print(next(error_bucket))
```
